### Section C - Edge cases
#### Q1: Array Shapes and Broadcasting


-

````python
import numpy as np

# Creating a column array with 3 values
a = np.array([[10], [20], [30]])

# Creating a row array with 3 values
b = np.array([[1, 2, 3]])

# NumPy broadcasts the arrays and performs addition
result = a + b

print(result)
print(result.shape)
````

**Answer:**

The output is:

```text
[[11 12 13]
 [21 22 23]
 [31 32 33]]
(3, 3)
```

The shape of `a` is `(3, 1)` and the shape of `b` is `(1, 3)`. After broadcasting, both arrays are treated as compatible with shape `(3, 3)`, so the result also has shape `(3, 3)`.

**Explanation:**
Broadcasting allows NumPy to perform operations on arrays with different shapes by automatically expanding their dimensions when they are compatible. Here, each value in `a` is added to all three values in `b`.

If `b = np.array([1, 2])`, broadcasting would not work because the dimensions are not compatible for this operation.

### Section D – Scenario-Based Questions

#### Q1. Student Assessment Data Cleaning Pipeline

**Scenario:
    
    
      **A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as `IT`, `it`, and `I.T.`, and marks stored as text[cite: 1, 3]. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type[cite: 1, 3]. State one risk of deleting missing values without investigating why they are missing[cite: 1, 3].

**Dataset:
    ** `student_assessment_dirty.csv`[cite: 1, 3]

### 1. Load the CSV file and inspect the dataset using head(), info(), and isnull().sum()

Loading `student_assessment_dirty.csv` and reviewing structural information, data types, and initial missing value counts across all columns.

````python
import io
import pandas as pd

# Creating in-memory dataset matching student_assessment_dirty.csv
csv_data = """Student_ID,Student_Name,Department,Assessment,Marks
S101,Asha,IT,Python,85
S102,Ravi,it,Python,78
S103,Neha,I.T.,Python,92
S104,Karan,HR,Python,
S105,Meera,HR,Python,Absent
S106,Arjun,Finance,Python,0
S107,Priya,finance,Python,67
S108,Rahul,IT,Python,75
S102,Ravi,it,Python,78
S109,Sneha,I.T.,Python,NA
S110,Varun,HR,Python,88 marks
S111,Diya,Finance,Python,91
S112,Aman,IT,Python,
S113,Nisha,HR,Python,64
S114,Kabir,Finance,Python,105"""

# To load directly from a file, replace io.StringIO with "student_assessment_dirty.csv"
df = pd.read_csv(io.StringIO(csv_data))

print("--- Head (First 5 Rows) ---")
print(df.head())

print("\n--- Info ---")
print(df.info())

print("\n--- Missing Values Count ---")
print(df.isnull().sum())
````

### 2. Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.

Cleaning whitespace and replacing inconsistent department entries (`it`, `I.T.` $\rightarrow$ `IT` and `finance` $\rightarrow$ `Finance`) to unify categories for aggregation.

````python
# Clean and replace department variations
df["Department"] = (
    df["Department"]
    .str.strip()
    .replace(
        {
            "it": "IT",
            "I.T.": "IT",
            "finance": "Finance",
            "hr": "HR",
        }
    )
)

print("Unique Departments after standardization:")
print(df["Department"].unique())
````

### 3. Identify and remove exact duplicate student assessment records.

Identifying exact repeated records across all columns and dropping duplicate entries.

````python
initial_count = len(df)

# Remove exact duplicate rows
df = df.drop_duplicates()

duplicates_removed = initial_count - len(df)
print(f"Number of duplicate records removed: {duplicates_removed}")
````

### 4. Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.

Extracting digits from text strings like `'88 marks'`, setting `'Absent'` students to a mark of `0`, and coercing invalid or blank entries (`'NA'`, empty strings) to Pandas `NaN`.

````python
# Convert to string and strip spaces
df["Marks"] = df["Marks"].astype(str).str.strip()

# Convert 'Absent' status to 0
df.loc[df["Marks"].str.lower() == "absent", "Marks"] = "0"

# Extract numeric digits from text such as '88 marks'
df["Marks"] = df["Marks"].str.extract(r"(\d+)")[0]

# Convert column to float numeric data type
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print("Marks column data type:", df["Marks"].dtype)
````

### 5. Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.

* **Distinction:   
     ** An `'Absent'` entry represents an active student who took zero marks, whereas a blank   or `'NA'` entry represents unrecorded data that requires administrative follow-up or imputation.
* **Risk of Deleting Missing Values:
     ** Deleting missing values directly introduces selection bias into the dataset, distorts descriptive statistics like the mean and median, and permanently removes valid student metadata.

````python
# Distinguish absent students (mark = 0) from unrecorded missing marks (NaN)
absent_students = df[df["Marks"] == 0]
missing_marks_students = df[df["Marks"].isnull()]

print("--- Absent Students (Marks = 0) ---")
print(absent_students[["Student_ID", "Student_Name", "Department", "Marks"]])

print("\n--- Unrecorded / Missing Marks (NaN) ---")
print(missing_marks_students[["Student_ID", "Student_Name", "Department", "Marks"]])
````

### 6. Identify marks outside the valid range of 0–100 and decide how they should be handled.

Marks outside $[0, 100]$ (such as `105` for `S114 Kabir`) represent data entry errors. They are flagged and replaced with `NaN` so they do not distort final statistical evaluations.

````python
# Find out-of-range marks (< 0 or > 100)
invalid_marks_mask = (df["Marks"] < 0) | (df["Marks"] > 100)
invalid_marks_count = invalid_marks_mask.sum()

# Convert out-of-range marks to NaN
df.loc[invalid_marks_mask, "Marks"] = None

print(f"Number of invalid marks identified and converted to NaN: {invalid_marks_count}")
````

### 7. Display the cleaned DataFrame and report the number of records removed, missing marks remaining, and invalid marks identified.

Final dataset view along with the final data cleaning summary report.

````python
# Calculate total remaining missing marks
missing_marks_remaining = df["Marks"].isnull().sum()

print("--- Cleaned DataFrame ---")
print(df)

print("\n--- Data Cleaning Summary Report ---")
print(f"1. Number of duplicate records removed: {duplicates_removed}")
print(f"2. Number of invalid marks identified (>100 or <0): {invalid_marks_count}")
print(f"3. Number of missing marks remaining (NaN): {missing_marks_remaining}")
````
